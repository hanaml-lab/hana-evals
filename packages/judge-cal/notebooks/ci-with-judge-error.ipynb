# A confidence interval that includes the judge's error

A judge says 72% of answers are acceptable. How sure should we be? The usual bootstrap resamples only the items, as if the judge were a human. Here we also resample the human labels the judge was calibrated on, so the interval widens when the judge disagrees with people or when there are few labels.


```python
import numpy as np
from judge_cal.calibrate import Calibrator, Scores
from judge_cal.report import summary

rng = np.random.default_rng(0)
# 300 human-labelled items: judge score 1-5 and whether a person accepted the answer
raw = rng.integers(1, 6, 300).astype(float)
labels = (rng.random(300) < (raw - 0.5) / 5).astype(float)
cal = Calibrator('isotonic', raw, labels).fit()

# 1,000 new items scored by the judge only
items = rng.integers(1, 6, 1000).astype(float)
scores = Scores(items, cal.predict(items), cal)
print(summary(scores, n_boot=1000))
```

Shrink the label set to 50 rows and the interval roughly doubles, while the judge-as-truth interval does not move.

```python
small = Calibrator('isotonic', raw[:50], labels[:50]).fit()
print(summary(Scores(items, small.predict(items), small), n_boot=1000))
```
